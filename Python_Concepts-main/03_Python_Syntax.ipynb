# 3. Python Syntax

**Syntax** means the *rules* for writing code correctly — just like grammar rules in English.

If we break the rules, Python gives an **error** and the code won't run.

## Rule 1: `print()` to show output

`print()` displays text or values on the screen.

Text (called a **string**) must be inside quotes `" "`.

```python
print("Welcome to Python")
print("Learning is fun")
```

## Rule 2: Comments with `#`

A **comment** is a note for humans. Python ignores it.

We use comments to explain our code.

```python
# This is a comment. Python will not run this line.
print("This line will run")  # comment can also be at the end of a line
```

## Rule 3: Indentation (spaces matter!)

Python uses **indentation** (spaces at the start of a line) to group code.

Other languages use `{ }`, but Python uses spaces. This is very important in Python.

```python
if 5 > 2:
    print("Five is greater than two")  # this line is indented (inside the if)
```

## Rule 4: Case-sensitive

Python treats uppercase and lowercase as **different**.

`Print` is NOT the same as `print`.

```python
name = "Ravi"   # small 'name'
Name = "Kumar"  # capital 'Name' -- this is a different variable
print(name)
print(Name)
```

## Key points to remember

- `print()` shows output; text goes inside quotes.
- `#` starts a comment (ignored by Python).
- Indentation (spaces) is required to group code.
- Python is case-sensitive (`name` ≠ `Name`).

```python
name ="Karthik"; print(name)
```
